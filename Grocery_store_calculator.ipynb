CASE STUDY:GROCERY STORE CALCULATION

```python
# Initialize variables
total_cost = 0
item_count = 0
items = []

# Start the grocery store calculator
print("Welcome to the Grocery Store Calculator!")
print("Enter 'done' when you are finished.")

while True:
    # Input item name
    item_name = input("Enter the name of the item (or 'done' to finish): ")

    if item_name.lower() == 'done':
        break

    # Input item price
    item_price = input(f"Enter the price of {item_name}: ")

    # Validate price input
    try:
        item_price = float(item_price)
    except ValueError:
        print("Invalid price. Please enter a numeric value.")
        continue

    # Input item quantity
    item_quantity = input(f"Enter the quantity of {item_name}: ")

    # Validate quantity input
    try:
        item_quantity = int(item_quantity)
        if item_quantity < 0:
            print("Quantity cannot be negative. Please enter a valid quantity.")
            continue
    except ValueError:
        print("Invalid quantity. Please enter a whole number.")
        continue

    # Calculate cost for the item
    item_cost = item_price * item_quantity
    total_cost += item_cost
    item_count += item_quantity

    # Store item details
    items.append((item_name, item_price, item_quantity, item_cost))

# Print the summary of items purchased
print("\n--- Grocery Store Receipt ---")
for item in items:
    print(f"{item[0]}: {item[1]:.2f} x {item[2]} = {item[3]:.2f}")

print(f"\nTotal items purchased: {item_count}")
print(f"Total cost: {total_cost:.2f}")
print("Thank you for shopping with us!")
```
